```python
import time
import matplotlib.pyplot as plt

def union_intersection(A, B):
    i = 0
    j = 0
    union = []
    intersection = []

    while i < len(A) and j < len(B):

        if A[i] < B[j]:
            union.append(A[i])
            i += 1

        elif A[i] > B[j]:
            union.append(B[j])
            j += 1

        else:
            union.append(A[i])
            intersection.append(A[i])
            i += 1
            j += 1

    while i < len(A):
        union.append(A[i])
        i += 1

    while j < len(B):
        union.append(B[j])
        j += 1

    return union, intersection


sizes = [100, 500, 1000, 2000, 5000, 10000]
times = []

for n in sizes:
    A = list(range(n))
    B = list(range(n))

    start = time.time()
    union_intersection(A, B)
    end = time.time()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Union and Intersection - Time Complexity")
plt.show()
```

```python
import time
import matplotlib.pyplot as plt

def search_rotated(A, target):

    low = 0
    high = len(A) - 1

    while low <= high:

        mid = (low + high) // 2

        if A[mid] == target:
            return mid

        if A[low] <= A[mid]:

            if A[low] <= target < A[mid]:
                high = mid - 1
            else:
                low = mid + 1

        else:

            if A[mid] < target <= A[high]:
                low = mid + 1
            else:
                high = mid - 1

    return -1


sizes = [100, 500, 1000, 2000, 5000, 10000, 20000, 50000]
times = []

for n in sizes:

    # Create rotated sorted array
    A = list(range(n // 2, n)) + list(range(n // 2))

    target = n - 1

    start = time.time()
    search_rotated(A, target)
    end = time.time()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Rotated Sorted Array Search - Time Complexity")
plt.show()
```

```python
import time
import matplotlib.pyplot as plt

def merge_sort(A):

    if len(A) <= 1:
        return A

    mid = len(A) // 2

    left = merge_sort(A[:mid])
    right = merge_sort(A[mid:])

    result = []

    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1

        else:
            result.append(right[j])
            j += 1

    while i < len(left):
        result.append(left[i])
        i += 1

    while j < len(right):
        result.append(right[j])
        j += 1

    return result


sizes = [100, 500, 1000, 2000, 5000, 10000]
times = []

for n in sizes:

    A = list(range(n, 0, -1))

    start = time.time()
    merge_sort(A)
    end = time.time()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Merge Sort - Time Complexity")
plt.show()
```

```python
import time
import matplotlib.pyplot as plt

def count_inversions(A):

    if len(A) <= 1:
        return A, 0

    mid = len(A) // 2

    left, count_left = count_inversions(A[:mid])
    right, count_right = count_inversions(A[mid:])

    result = []
    i = 0
    j = 0

    count = count_left + count_right

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1

        else:
            result.append(right[j])

            count += len(left) - i

            j += 1

    while i < len(left):
        result.append(left[i])
        i += 1

    while j < len(right):
        result.append(right[j])
        j += 1

    return result, count


sizes = [100, 500, 1000, 2000, 5000, 10000]
times = []

for n in sizes:

    # Reverse order gives many inversions
    A = list(range(n, 0, -1))

    start = time.time()
    count_inversions(A)
    end = time.time()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Count Inversions - Time Complexity")
plt.show()
```
