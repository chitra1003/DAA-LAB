```python
import matplotlib.pyplot as plt

n = [10, 20, 30, 40, 50]
t = [x**2 for x in n]

plt.plot(n, t, 'o-')
plt.xlabel("Input Size")
plt.ylabel("Time Complexity")
plt.title("Relative Sorting - O(n²)")
plt.grid()
plt.show()
```
