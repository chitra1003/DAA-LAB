```python

import matplotlib.pyplot as plt
n = list(range(1, 101))
y = [v*v for v in n]
plt.plot(n, y)
plt.xlabel("Vertices")
plt.ylabel("Time Complexity")
plt.title("Dijkstra: O(V^2)")
plt.show()
```

```python

import matplotlib.pyplot as plt
n = list(range(1, 101))
y = n
plt.plot(n, y)
plt.xlabel("Number of Prices")
plt.ylabel("Time Complexity")
plt.title("Stock Buy/Sell: O(n)")
plt.show()
```
