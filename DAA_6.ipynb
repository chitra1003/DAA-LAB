```python
import random, time
import matplotlib.pyplot as plt

def counting_sort(a):
    count = [0] * (max(a) + 1)
    for x in a:
        count[x] += 1
    result = []
    for i in range(len(count)):
        result += [i] * count[i]
    return result

sizes = [100, 500, 1000, 2000, 5000, 10000]
times = []

for n in sizes:
    a = [random.randint(0, 1000) for _ in range(n)]
    start = time.time()
    counting_sort(a)
    times.append(time.time() - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Time (seconds)")
plt.title("Counting Sort Time Complexity")
plt.show()
```

```python
import random, time
import matplotlib.pyplot as plt

def bucket_sort(a):
    n = len(a)
    buckets = [[] for _ in range(n)]

    for x in a:
        buckets[int(x * n)].append(x)

    result = []
    for b in buckets:
        b.sort()
        result += b
    return result

sizes = [100, 500, 1000, 2000, 5000, 10000]
times = []

for n in sizes:
    a = [random.random() for _ in range(n)]
    start = time.time()
    bucket_sort(a)
    times.append(time.time() - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Time (seconds)")
plt.title("Bucket Sort Time Complexity")
plt.show()
```

```python
import random, time
import matplotlib.pyplot as plt

def at_most(a, k):
    left = count = 0
    freq = {}

    for right in range(len(a)):
        freq[a[right]] = freq.get(a[right], 0) + 1

        while len(freq) > k:
            freq[a[left]] -= 1
            if freq[a[left]] == 0:
                del freq[a[left]]
            left += 1

        count += right - left + 1

    return count

def exact_k(a, k):
    return at_most(a, k) - at_most(a, k - 1)

sizes = [100, 500, 1000, 2000, 5000, 10000]
times = []

for n in sizes:
    a = [random.randint(1, 20) for _ in range(n)]
    start = time.time()
    exact_k(a, 5)
    times.append(time.time() - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Time (seconds)")
plt.title("Exactly K Distinct Subarrays")
plt.show()
```

```python
import random, time
import heapq
import matplotlib.pyplot as plt

def minimize_sum(a, k):
    heap = [-x for x in a]
    heapq.heapify(heap)

    for _ in range(k):
        x = -heapq.heappop(heap)
        x = (x + 1) // 2
        heapq.heappush(heap, -x)

    return -sum(heap)

sizes = [100, 500, 1000, 2000, 5000, 10000]
times = []

for n in sizes:
    a = [random.randint(1, 1000) for _ in range(n)]
    k = n
    start = time.time()
    minimize_sum(a, k)
    times.append(time.time() - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Time (seconds)")
plt.title("Minimize Array Sum Time Complexity")
plt.show()
```
