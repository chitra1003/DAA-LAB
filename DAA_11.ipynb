```python

import matplotlib.pyplot as plt
n = list(range(1, 101))
y = [v**3 for v in n]
plt.plot(n, y)
plt.xlabel("Vertices")
plt.ylabel("Time Complexity")
plt.title("Bellman-Ford: O(VE)")
plt.show()
```
