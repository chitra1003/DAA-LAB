```python
import time
import matplotlib.pyplot as plt

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key


sizes = [100, 200, 400, 800, 1600, 3200]
times = []

for n in sizes:
    arr = list(range(n, 0, -1))   # Worst-case input

    start = time.perf_counter()
    insertion_sort(arr)
    end = time.perf_counter()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Time (seconds)")
plt.title("Insertion Sort Time Complexity")
plt.grid()
plt.show()
```

```python
import time
import random
import matplotlib.pyplot as plt

def dutch_national_flag(arr):
    low = 0
    mid = 0
    high = len(arr) - 1

    while mid <= high:
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1

        elif arr[mid] == 1:
            mid += 1

        else:
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1


sizes = [100, 200, 400, 800, 1600, 3200]
times = []

for n in sizes:
    arr = [random.randint(0, 2) for _ in range(n)]

    start = time.perf_counter()
    dutch_national_flag(arr)
    end = time.perf_counter()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Time (seconds)")
plt.title("Dutch National Flag Time Complexity")
plt.grid()
plt.show()
```

```python
import time
import random
import matplotlib.pyplot as plt

def moore_voting(arr):
    candidate = None
    count = 0

    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1

    # Verification
    frequency = 0
    for x in arr:
        if x == candidate:
            frequency += 1

    if frequency > len(arr) // 2:
        return candidate

    return None


sizes = [100, 200, 400, 800, 1600, 3200]
times = []

for n in sizes:
    # Create input with a majority element
    arr = [1] * (n // 2 + 1)

    while len(arr) < n:
        arr.append(random.randint(2, 5))

    random.shuffle(arr)

    start = time.perf_counter()
    moore_voting(arr)
    end = time.perf_counter()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Time (seconds)")
plt.title("Moore's Voting Algorithm Time Complexity")
plt.grid()
plt.show()
```
