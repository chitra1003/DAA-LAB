```python
import time
import matplotlib.pyplot as plt

def max_product_subarray(arr):
    max_product = arr[0]
    max_here = arr[0]
    min_here = arr[0]

    for i in range(1, len(arr)):
        if arr[i] < 0:
            max_here, min_here = min_here, max_here

        max_here = max(arr[i], max_here * arr[i])
        min_here = min(arr[i], min_here * arr[i])
        max_product = max(max_product, max_here)

    return max_product


sizes = [1000, 2000, 3000, 4000, 5000]
times = []

for n in sizes:
    arr = [2] * n

    start = time.perf_counter()
    max_product_subarray(arr)
    end = time.perf_counter()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Maximum Product Subarray - O(n)")
plt.grid()
plt.show()
```

```python
import time
import matplotlib.pyplot as plt
import sys

def set_recursion_limit(limit):
    original_limit = sys.getrecursionlimit()
    sys.setrecursionlimit(limit)
    return original_limit

def restore_recursion_limit(original_limit):
    sys.setrecursionlimit(original_limit)

def partition(arr, low, high):
    pivot = arr[high]
    i = low

    for j in range(low, high):
        if arr[j] > pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[high] = arr[high], arr[i]
    return i


def quick_select(arr, low, high, k):
    if low <= high:
        pos = partition(arr, low, high)

        if pos == k - 1:
            return arr[pos]
        elif pos > k - 1:
            return quick_select(arr, low, pos - 1, k)
        else:
            return quick_select(arr, pos + 1, high, k)


sizes = [1000, 2000, 3000, 4000, 5000]
times = []

original_recursion_limit = set_recursion_limit(2 * max(sizes))

for n in sizes:
    arr = list(range(n, 0, -1))
    k = n // 2

    start = time.perf_counter()
    quick_select(arr, 0, n - 1, k)
    end = time.perf_counter()

    times.append(end - start)

restore_recursion_limit(original_recursion_limit)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Quick Select - Worst-case O(n^2) with sorted input")
plt.grid()
plt.show()
```

```python
import time
import matplotlib.pyplot as plt

def fibonacci(n):
    if n <= 1:
        return n

    a = 0
    b = 1

    for i in range(2, n + 1):
        c = a + b
        a = b
        b = c

    return b


sizes = [1000, 2000, 3000, 4000, 5000]
times = []

for n in sizes:
    start = time.perf_counter()
    fibonacci(n)
    end = time.perf_counter()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Fibonacci - O(n)")
plt.grid()
plt.show()
```

```python
import time
import matplotlib.pyplot as plt

def smallest_missing_positive(arr):
    n = len(arr)

    for i in range(n):
        while 1 <= arr[i] <= n and arr[arr[i] - 1] != arr[i]:
            arr[arr[i] - 1], arr[i] = arr[i], arr[arr[i] - 1]

    for i in range(n):
        if arr[i] != i + 1:
            return i + 1

    return n + 1


sizes = [1000, 2000, 3000, 4000, 5000]
times = []

for n in sizes:
    arr = list(range(1, n + 1))
    arr[-1] = -1

    start = time.perf_counter()
    smallest_missing_positive(arr)
    end = time.perf_counter()

    times.append(end - start)

plt.plot(sizes, times, marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Smallest Missing Positive - O(n)")
plt.grid()
plt.show()
```
